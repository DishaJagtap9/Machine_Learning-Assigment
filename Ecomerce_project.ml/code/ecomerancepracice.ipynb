```python
from scipy.stats import chi2_contingency

categorical_col = ['Gender', 'Membership', 'Preferred_Channel',
                   'Country', 'Payment_Method']

for col in categorical_col:
    table = pd.crosstab(df[col], df['Purchased'])
    
    chi2, p, dof, expected = chi2_contingency(table)
    
    print(col, "p-value:", p)
```

```python
from scipy.stats import pearsonr

for col in numerical_col:
    corr,p_value=pearsonr(df[col],df['Purchased'])

    print(f"\n{col}")
    print(f"Correlation: {corr:.4f}")
    print(f"P_value{p_value:.4f}")

    if p_value<0.05:
        print("it is correlated")
    else:
        print("it is not correlated")
```

```python
df=df.drop('Customer_ID',axis=1)
```

```python
df['Gender']=df['Gender'].str.strip().str.lower()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Gender']=le.fit_transform(df['Gender'])
```

```python
print(df.head())
```

```python
df=pd.get_dummies(df,columns=['Membership', 'Preferred_Channel',
                   'Country', 'Payment_Method'])
```

```python
df.head()
```

```python
df=df.astype(int)
```

```python
print(df.head())
print(df.tail())
```

```python
df.duplicated().sum()
```

```python
df=df.drop_duplicates()
```

```python
X=df.drop('Purchased',axis=1)
y=df['Purchased']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.preprocessing import StandardScaler
scalar=StandardScaler()
x_train_scaled=scalar.fit_transform(X_train)
x_test_scaled=scalar.transform(X_test)
```

```python
from sklearn.linear_model import LinearRegression
model1=LinearRegression()
```

```python
model1.fit(x_train_scaled,y_train)
```

```python
predict_y1=model1.predict(x_test_scaled)
```

```python
from sklearn.metrics import r2_score,root_mean_squared_error,mean_absolute_error
acc1= r2_score(y_test,predict_y1)
print(acc1)

```

#Logistic Regression


```python
from sklearn.linear_model import LogisticRegression
model2=LogisticRegression(random_state=42)
model2.fit(x_train_scaled,y_train)
```

```python
y_predict=model2.predict(x_test_scaled)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
acc2=accuracy_score(y_test,y_predict)
print(acc2)
confusion_matrix=confusion_matrix(y_test,y_predict)
print(confusion_matrix)
classification_report=classification_report(y_test,y_predict)
print(classification_report)
```

```python
#Knn
```

```python
from sklearn.neighbors import KNeighborsClassifier
model4=KNeighborsClassifier(n_neighbors=3)
model4.fit(x_train_scaled,y_train)
y4=model4.predict(x_test_scaled)
y5=model4.predict(x_train_scaled)
print("testing",accuracy_score(y_test,y4))
print("training",accuracy_score(y_train,y5))
```

#Random Forest

```python
from sklearn.ensemble import RandomForestClassifier
model3=RandomForestClassifier(random_state=42)
model3.fit(x_train_scaled,y_train)
y3=model3.predict(x_test_scaled)
print(accuracy_score(y_test,y3))
```
