```python
df['Overtime']=df['Overtime'].replace('Y','yes')
df['Overtime']=df['Overtime'].replace('y','yes')
df['Overtime']=df['Overtime'].replace('N','no')
df['Overtime']=df['Overtime'].replace('n','no')
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
df['Overtime']=le.fit_transform(df['Overtime'])
```

```python
df['Attrition']=le.fit_transform(df['Attrition'])
```

```python
df.head()
```

```python
df=pd.get_dummies(df,columns=['Job_Role','Department','Education','Marital_Status'],dtype=int)
```

```python
df.isnull().sum()
```

```python
print(df.head())
```

```python
X=df.drop('Attrition',axis=1)
y=df['Attrition']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
X_train, X_test, y_train, y_test = train_test_split(
       X, y, test_size=0.20, random_state=42)
```

```python
X_train.isnull().sum()
```

```python
scalar=StandardScaler()
X_train_scaled=scalar.fit_transform(X_train)
X_test_scaled=scalar.transform(X_test)
```

```python
X_train.isnull().sum()
```

```python
from sklearn.linear_model import LogisticRegression
model1=LogisticRegression()
model1.fit(X_train_scaled,y_train)
```

```python
y1_predic=model1.predict(X_test_scaled)
```

```python
from sklearn.metrics import accuracy_score
print("LogisticRegression:",accuracy_score(y_test,y1_predic))
```

```python
from sklearn.ensemble import RandomForestClassifier
model2=RandomForestClassifier()
model2.fit(X_train_scaled,y_train)
```

```python
y2_predic=model2.predict(X_test_scaled)
print("random forest",accuracy_score(y_test,y2_predic))
```

```python
from xgboost import XGBClassifier
model3=XGBClassifier()
model3.fit(X_train_scaled,y_train)
```

```python
y3_predict=model3.predict(X_test_scaled)
print("Xgboost accuracy",accuracy_score(y_test,y3_predict))
```

```python
print("---"*55)
print("Accuracy")
print("---"*55)
print("LogisticRegression:",accuracy_score(y_test,y1_predic))
print("random forest",accuracy_score(y_test,y2_predic))
print("Xgboost accuracy",accuracy_score(y_test,y3_predict))
print("---"*55)
print("---"*55)
```

```python
lr=LogisticRegression()
lr.fit(X_train_scaled,y_train)
```

```python
predict_result=lr.predict(X_test_scaled)
print(accuracy_score(y_test,predict_result))
```

```python
from sklearn.model_selection import GridSearchCV
param_grid ={
    'C': [0.01, 0.1, 1, 10],
    'solver': ['liblinear', 'lbfgs'],
    'max_iter': [100, 200, 500]
}
grid=GridSearchCV(
    estimator=lr,
    param_grid=param_grid,
    scoring='accuracy',
    cv=5,
    n_jobs=-1
)
grid.fit(X_train_scaled, y_train)
```

```python
print("best_parameter:",grid.best_params_)
print("best score:",grid.best_score_)
```
